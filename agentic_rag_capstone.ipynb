```python
import os
import numpy as np
from sentence_transformers import SentenceTransformer
from langchain_core.tools import tool
from langchain_groq import ChatGroq
from langgraph.prebuilt import create_react_agent

# 1. Read key automatically from file
with open("groq_key.txt", "r") as f:
    os.environ["GROQ_API_KEY"] = f.read().strip()

# 2. Local Sentence Transformer (Runs offline in Colab memory)
embed_model = SentenceTransformer("all-MiniLM-L6-v2")

docs = [
    "Samsung Innovation Campus policy: Attendance above 90% is strictly mandatory to receive certification.",
    "Travel reimbursement rate: Students and employees receive 18 rupees per kilometer for official travel visits.",
    "Capstone guidelines: Final project submissions require a GitHub repository, an architecture diagram, and a live working demo link."
]
doc_embeddings = embed_model.encode(docs)

# 3. Define Tools
@tool
def search_policy_docs(query: str) -> str:
    """Searches official guidelines and policy documentation for rules, rates, and criteria."""
    q_emb = embed_model.encode([query])
    scores = np.dot(doc_embeddings, q_emb.T).flatten()
    best_idx = int(np.argmax(scores))
    return docs[best_idx]

@tool
def multiply(a: float, b: float) -> float:
    """Multiplies two numbers together for exact numerical calculations."""
    return a * b

tools = [search_policy_docs, multiply]

# 4. Model Setup
llm = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0.0
)

# 5. Execute Agent
agent = create_react_agent(llm, tools)

query = "What are the submission requirements for the final capstone project?"

print("User Query:", query)

response = agent.invoke({"messages": [("user", query)]})
print("\nAgent Response:\n" + response["messages"][-1].content)
```
